<font size=5>회전 초밥집에서는 담긴 접시의 색에 따라 초밥의 가격이 다르게 계산된다.<br>다음을 참고하여 회전 초밥의 비용을 영수증 형식으로 출력하는 프로그램을 완성해 보자.<br>

<br><br><br>

##입력예시


```
빨간색 접시의 개수를 입력하세요: 2
노란색 접시의 개수를 입력하세요: 3
파란색 접시의 개수를 입력하세요: 1
초록색 접시의 개수를 입력하세요: 2
```


##출력예시
```
========= 회전 초밥 영수증 =======
 접시     단가     개수       합계
빨간색   1,000       2       2,000
노란색   2,000       3       6,000
파란색   3,000       1       3,000
초록색   4,000       2       8,000
총 가격                     19,000
==================================
```



<font size=5>
입력 = 4개<br>
변수 = n개<br>
필요한기능
input(),print(),이스케이프 시퀀스, 계산(합계)

````python
red = int(input("빨간색 접시의 개수를 입력하세요:"))#빨간색 접시의 계수를 저장하는 변수
red_price = 1000 #빨간색 접시의 단가
red_price_total = red*red_price #빨간색 접시의 가격 총합
sushi_price_total = red_price_total#전체 가격 총합
print("======= 회전 초밥 영수증 =======")
print("접시\t단가\t개수\t합계")
print(f"빨간색\t{red_price}\t{red}\t{red_price_total}")
print(f"총 가격\t\t\t{sushi_price_total}")
````

<font size=5>적금은 대표적인 금융 상품으로 일정 기간 금액을 납부한 후 이자와 함께 돌려받는 것을 말한다. 다음의 실행 결과를 보고 ‘적금 계산기’ 프로그램을 만들어 보자.
##입력예시

```
월 납입액을 입력하세요 (원): 100000
연 이자율을 입력하세요 (%): 3.75
적금 기간을 입력하세요 (월): 24
```

##출력예시
```
적금 기간 납입한 총 원금: 2400000 원
적금 기간 발생한 총 이자: 7500 원
적금 만기 시 받을 총금액: 2407500 원
```

1개월 10,000원 = 1,000
11개월 10,000원 = 83

<font size=8>
납입 원금 = 10,000<br>
연 이자율 = 10%<br>
납입기간 = 12개월<br>
10000 * 0.1 * 12
원금 X 연이율 X 기간(년)

````python
a = int(input("월 납입액을 입력하세요 (원):"))
b = int(input("연 이자율을 입력하세요 (%):")) / 100
c = int(input("적금 기간을 입력하세요 (월):"))
print(f"납입 원금 = {a}\n연 이자율 = {b}%\n납입 기간 = {c}개월")
````
